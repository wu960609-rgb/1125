**測試一**

```python
# ========================================== #
# HW09 兩場活動報名整理器  #
# ========================================== #

# --- 1. 原始測試資料與活動參數設定 ---
event_a_name = "桌遊體驗"
event_a_price = 120.00
event_a_raw = " amy , ben, amy, cora "

event_b_name = "電影欣賞"
event_b_price = 80.00
event_b_raw = " ben, dora , eric, ben "

# 顯示簡短名稱，用於符合「桌遊整理串列」等標籤
short_name_a = "桌遊"
short_name_b = "電影"

# --- 2. 資料清理與轉換 (字串處理 -> 串列) ---
clean_a = event_a_raw.lower().replace(" ", "")
list_a = clean_a.split(",")

clean_b = event_b_raw.lower().replace(" ", "")
list_b = clean_b.split(",")

# --- 3. 轉換為集合去重，並轉回排序後的串列 ---
set_a = set(list_a)
set_b = set(list_b)

list_a_unique = sorted(list(set_a))
list_b_unique = sorted(list(set_b))

# --- 4. 集合運算與名單轉換 ---
# 共同參加者 (交集)
both_set = set_a & set_b
both_list = sorted(list(both_set))

# 只參加者 (差集)
only_a_set = set_a - set_b
only_a_list = sorted(list(only_a_set))

# 只參加者 (差集)
only_b_set = set_b - set_a
only_b_list = sorted(list(only_b_set))

# 全部參加者 (聯集)
all_set = set_a | set_b
all_list = sorted(list(all_set))

# --- 5. 人數與人次計算 ---
unique_count_a = len(set_a)
unique_count_b = len(set_b)
both_count = len(both_set)
total_unique_people = len(all_set)

# 依預期結果定義，報名人次為兩邊「去重後人數」之和 (3 + 3 = 6 人次)
total_registrations = unique_count_a + unique_count_b

# 通知名單前兩位
notify_list = all_list[:2]

# --- 6. 費用計算 ---
fee_a = unique_count_a * event_a_price
fee_b = unique_count_b * event_b_price
total_fee = fee_a + fee_b

# --- 7. 格式化輸出結果 ---
print("HW09 活動報名摘要")
print(f"活動：{event_a_name}，每人 {event_a_price:.2f} 元")
print(f"活動：{event_b_name}，每人 {event_b_price:.2f} 元")
print(f"{short_name_a}整理串列：{list_a}")
print(f"{short_name_b}整理串列：{list_b}")
print(f"{short_name_a}去重名單：{list_a_unique}")
print(f"{short_name_b}去重名單：{list_b_unique}")
print(f"{short_name_a}人數：{unique_count_a} 人")
print(f"{short_name_b}人數：{unique_count_b} 人")
print(f"共同參加者：{both_list}")
print(f"共同人數：{both_count} 人")
print(f"只參加{short_name_a}者：{only_a_list}")
print(f"只參加{short_name_b}者：{only_b_list}")
print(f"全部參加者：{all_list}")
print(f"不重複總人數：{total_unique_people} 人")
print(f"報名人次：{total_registrations} 人次")
print(f"通知名單前兩位：{notify_list}")
print(f"{short_name_a}應收：{fee_a:.2f} 元")
print(f"{short_name_b}應收：{fee_b:.2f} 元")
print(f"總應收：{total_fee:.2f} 元")
```
